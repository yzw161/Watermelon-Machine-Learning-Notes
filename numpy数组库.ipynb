```python

import numpy as np 
```

```python
# 递增数组 
arr1 = np.arange(10)    
print(arr1)
```

```python
# 递增数组 
arr2 = np.arange(10,20)   
print(arr2)
```

```python
# 递增数组 
arr3 = np.arange(1,21,2)  # 从 1 开始，到 21 之前结束，步长为2 
print(arr3)
```

```python
# 全0数组 
arr1 = np.zeros( 3 )           
print(arr1) 
```

```python
# 全1数组 
arr2 = np.ones( (1,3) )         
print(arr2)
```

```python
# 全3.14数组 
arr3 = 3.14 * np.ones( (2,3) )    
print(arr3) 
```

```python
# 0-1 均匀分布的浮点型随机数组 
arr1 = np.random.random( 5 )             
print( arr1 )
```

```python
# 整数型随机数组 
arr2 = np.random.randint( 10,100,(1,15) )    
print( arr2 ) 
```

```python
# 服从正态分布的随机数组 
arr3 = np.random.normal( 0,1,(2,3) )      
print( arr3 ) 
```

```python
# 创建向量 
arr1 = np.arange( 1,10 ) 
print( arr1 )
```

```python
# 访问元素 
print( arr1[3] )     # 正着访问 
print( arr1[-1] )    # 倒着访问
```

```python
# 修改数组元素 
arr1[3] = 100; 
print( arr1 ) 
```

```python
# 创建矩阵 
arr2 = np.array( [ [1,2,3],[4,5,6] ] ) 
print(arr2) 
```

```python
# 访问元素 
print( arr2[0,2] ) 
print( arr2[1,-2] ) 
```

```python
# 修改元素 
arr2[1,1] = 100.9 
print( arr2 ) 
```

```python
# 创建向量 
arr1 = np.arange(0,90,10) 
print(arr1) 
```

```python
# 花式索引 
print( arr1[ [0,2] ] )
```
